```python
import tensorflow as tf 
from tensorflow.keras.models import load_model 
import pickle
import pandas as pd 
import numpy as np 
import warnings 
warnings.filterwarnings('ignore')
```

```python
## load the trained model and preprocessor pickel file 
model = load_model('model.h5') 

# load the preprocessor file 
with open('preprocessor.pkl','rb') as file: 
    preprocessor = pickle.load(file)
```

```python
### example of some input data 
new_customer_data = { 
    'CreditScore' : 600,
    'Geography' :'France',
    'Gender' : 'Male',
    'Age' :40,
    'Tenure' : 3,
    'Balance' : 60000,
    'NumOfProducts': 2,
    'HasCrCard' : 1,
    'IsActiveMember':1, 
    'EstimatedSalary':50000
}
```

```python

#converting dictionary into the data frame
new_customer = pd.DataFrame([new_customer_data])

## then applying preprocessing step
new_customer_processed = preprocessor.transform(new_customer) 
new_customer_processed = pd.DataFrame(new_customer_processed,columns=preprocessor.get_feature_names_out()) 
new_customer_processed
```

```python
new_customer
```

```python

## prediction 
prediction = model.predict(new_customer_processed)

probability = prediction[0][0]

print("Churn Probability:", probability)
```

```python
## setting the threshold value
if probability >= 0.5:
    print("Customer is likely to churn")
else:
    print("Customer is not likely to churn")
```
